```python
# AF1 - Linear Regression: Predicting Monthly Spend
# Import the required libraries
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
```

```python
# 1. Data loading
data = {
    "customer_id": [1, 2, 3, 4, 5, 6, 7, 8],
    "monthly_visits": [5, 8, 4, 10, 7, 6, 9, 3],
    "time_in_store_min": [20, 35, 15, 40, 25, 22, 38, 12],
    "monthly_spend_usd": [150, 245, 110, 300, 210, 180, 275, 90],
}
df = pd.DataFrame(data)
df
```

```python
# 2. Exploratory analysis
# Descriptive statistics and correlation between variables
print("Description:")
print(df.describe())
print()
print("Correlation matrix:")
print(df[["monthly_visits", "time_in_store_min", "monthly_spend_usd"]].corr())
```

```python
# 3. Model training
# Predictor variables (X) and target variable (y)
X = df[["monthly_visits", "time_in_store_min"]]
y = df["monthly_spend_usd"]

model = LinearRegression()
model.fit(X, y)

w0=model.intercept_
w1=model.coef_[0]
w2=model.coef_[1]

print("W0: ", model.intercept_)
print("W1: ", model.coef_[0])
print("W2: ", model.coef_[1])
print("Coefficients:", dict(zip(X.columns, model.coef_)))
print("Intercept:", model.intercept_)

#manual prediction
new_visits= 8
new_time=25
prediction = w0 + w1 * new_visits + w2 * new_time

print("Manual Prediction: ", prediction)

#Library Prediction
new_customer=[[8,25]]
prediction=  model.predict(pd.DataFrame(new_customer, columns=["monthly_visits", "time_in_store_min"]))
print("Linear Regression Prediction: ", prediction)
```

```python
# 4. Model evaluation
df["predicted_spend_usd"] = model.predict(X).round(2)
mae = mean_absolute_error(y, df["predicted_spend_usd"])

print(f"MAE (Mean Absolute Error): {mae:.2f} USD")
df[["customer_id", "monthly_spend_usd", "predicted_spend_usd"]]
```

**¿Qué representa la matriz de coeficientes que obtuviste?:**

* Sin embargo, dado que este MAE se calculó sobre los datos que fueron usados para entrenar el modelo (actualmente no existe entrenamiento/testing) y que el dataset utilizado es demasiado pequeño (8 valores), el error bajo refleja que el modelo se ajusta a estos datos específicos, pero podría no garantizar el mismo desempeño con datos de clientes nuevos.
